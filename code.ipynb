```python
#AU2220195-VarunKotwani
#AU2420189-Jiya Thakkar
#AU2210484-Jeeya Gohil
#AU2210189-Dhyey Patel


# ===============================================================
# 1️⃣ Install & Import Dependencies
# ===============================================================
!pip install -q transformers torch scikit-learn pandas numpy tqdm accelerate

import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader
from torch.optim import AdamW
from transformers import AutoTokenizer, AutoModel
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from tqdm.auto import tqdm
import joblib
import os
from torch.cuda.amp import GradScaler, autocast

plt.style.use("ggplot")

# ===============================================================
# 2️⃣ Config & Device
# ===============================================================
DEVICE = "cuda" if torch.cuda.is_available() else "cpu"
MODEL_NAME = "distilbert-base-uncased"
print("🚀 Using device:", DEVICE)

# ===============================================================
# 3️⃣ Load Dataset
# ===============================================================
path = "/kaggle/input/revies-analysis"  # adjust if needed
train_df = pd.read_excel(path + "/training_data.xlsx")
test_df  = pd.read_excel(path + "/testing_data.xlsx")

# ===============================================================
# 4️⃣ Clean Sentiments
# ===============================================================
def simplify_sentiment(label):
    label = str(label).lower()
    if 'positive' in label:
        return 'Positive'
    elif 'negative' in label:
        return 'Negative'
    elif 'neutral' in label:
        return 'Neutral'
    else:
        return 'Neutral'

train_df['predicted_sentiments'] = train_df['predicted_sentiments'].apply(simplify_sentiment)
test_df['predicted_sentiments']  = test_df['predicted_sentiments'].apply(simplify_sentiment)

# ===============================================================
# 5️⃣ Feature Prep
# ===============================================================
meta_cols = [
    'Actuation_Score','Awareness_Score','Connectivity_Score',
    'Dynamism_Score','Novelty_Score','Personality_Score'
]

train_df = train_df.dropna(subset=['clean_review', 'predicted_sentiments'])
test_df  = test_df.dropna(subset=['clean_review', 'predicted_sentiments'])

le = LabelEncoder()
train_df['label'] = le.fit_transform(train_df['predicted_sentiments'])
test_df['label']  = le.transform(test_df['predicted_sentiments'])

scaler = StandardScaler()
train_df[meta_cols] = scaler.fit_transform(train_df[meta_cols].fillna(0))
test_df[meta_cols]  = scaler.transform(test_df[meta_cols].fillna(0))

# ===============================================================
# 6️⃣ Pre-tokenize Texts (⚡ Major Speed Boost)
# ===============================================================
tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)

def tokenize_texts(df):
    return tokenizer(
        df["clean_review"].astype(str).tolist(),
        truncation=True,
        padding="max_length",
        max_length=128,
        return_tensors="pt"
    )

train_enc = tokenize_texts(train_df)
test_enc  = tokenize_texts(test_df)

# ===============================================================
# 7️⃣ Dataset & DataLoader
# ===============================================================
class ReviewDataset(Dataset):
    def __init__(self, encodings, metas, labels):
        self.encodings = encodings
        self.metas = metas
        self.labels = labels

    def __len__(self):
        return len(self.labels)

    def __getitem__(self, idx):
        return {
            "input_ids": self.encodings["input_ids"][idx],
            "attention_mask": self.encodings["attention_mask"][idx],
            "meta": torch.tensor(self.metas[idx], dtype=torch.float),
            "label": torch.tensor(self.labels[idx], dtype=torch.long),
        }

train_ds = ReviewDataset(train_enc, train_df[meta_cols].values, train_df["label"].values)
test_ds  = ReviewDataset(test_enc,  test_df[meta_cols].values,  test_df["label"].values)

train_loader = DataLoader(train_ds, batch_size=64, shuffle=True, num_workers=4, pin_memory=True)
test_loader  = DataLoader(test_ds, batch_size=128, shuffle=False, num_workers=4, pin_memory=True)

# ===============================================================
# 8️⃣ Hybrid Transformer Model
# ===============================================================
class HybridTransformer(nn.Module):
    def __init__(self, model_name, meta_dim, num_labels, freeze_base=True):
        super().__init__()
        self.transformer = AutoModel.from_pretrained(model_name)
        hidden_size = self.transformer.config.hidden_size
        
        # Optional: freeze base layers for faster fine-tuning
        if freeze_base:
            for name, param in self.transformer.named_parameters():
                if "layer.4" not in name and "layer.5" not in name:
                    param.requires_grad = False
        
        self.meta_net = nn.Sequential(
            nn.Linear(meta_dim, 32),
            nn.ReLU(),
            nn.Dropout(0.2)
        )
        
        self.classifier = nn.Sequential(
            nn.Linear(hidden_size + 32, 128),
            nn.ReLU(),
            nn.Dropout(0.3),
            nn.Linear(128, num_labels)
        )
    
    def forward(self, input_ids, attention_mask, meta):
        outputs = self.transformer(input_ids=input_ids, attention_mask=attention_mask)
        text_feat = outputs.last_hidden_state[:, 0, :]  # CLS token
        meta_feat = self.meta_net(meta)
        combined = torch.cat((text_feat, meta_feat), dim=1)
        return self.classifier(combined)

model = HybridTransformer(MODEL_NAME, len(meta_cols), len(le.classes_), freeze_base=True).to(DEVICE)

# ===============================================================
# 9️⃣ Training Loop with Mixed Precision (AMP)
# ===============================================================
optimizer = AdamW(model.parameters(), lr=2e-5)
criterion = nn.CrossEntropyLoss()
EPOCHS = 3

scaler = GradScaler()

for epoch in range(EPOCHS):
    model.train()
    total_loss = 0
    loop = tqdm(train_loader, desc=f"Epoch {epoch+1}/{EPOCHS}", leave=False)
    
    for batch in loop:
        optimizer.zero_grad()
        input_ids = batch["input_ids"].to(DEVICE)
        attention_mask = batch["attention_mask"].to(DEVICE)
        meta = batch["meta"].to(DEVICE)
        labels = batch["label"].to(DEVICE)

        with autocast():
            outputs = model(input_ids, attention_mask, meta)
            loss = criterion(outputs, labels)

        scaler.scale(loss).backward()
        scaler.step(optimizer)
        scaler.update()
        total_loss += loss.item()

        loop.set_postfix(loss=loss.item())

    print(f"Epoch {epoch+1}/{EPOCHS} | Training Loss: {total_loss/len(train_loader):.4f}")

# ===============================================================
# 🔟 Evaluation
# ===============================================================
model.eval()
preds, trues = [], []

with torch.no_grad():
    for batch in tqdm(test_loader, desc="Evaluating"):
        input_ids = batch["input_ids"].to(DEVICE)
        attention_mask = batch["attention_mask"].to(DEVICE)
        meta = batch["meta"].to(DEVICE)
        labels = batch["label"].to(DEVICE)

        with autocast():
            outputs = model(input_ids, attention_mask, meta)

        preds.extend(outputs.argmax(1).cpu().numpy())
        trues.extend(labels.cpu().numpy())

acc = accuracy_score(trues, preds)
print(f"\n🎯 Turbo Hybrid Accuracy: {acc:.4f}")
print("\n📊 Classification Report:\n", classification_report(trues, preds, target_names=le.classes_))

cm = confusion_matrix(trues, preds)
sns.heatmap(cm, annot=True, fmt="d", cmap="Blues",
            xticklabels=le.classes_, yticklabels=le.classes_)
plt.title("Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.show()

# ===============================================================
# 💾 Save Model + Tokenizer + Scaler + Encoder
# ===============================================================
torch.save(model.state_dict(), "hybrid_transformer_turbo.pt")
tokenizer.save_pretrained("./tokenizer")
joblib.dump(scaler, "meta_scaler.pkl")
joblib.dump(le, "label_encoder.pkl")

print("✅ All components saved successfully (Turbo Version)!")
```
